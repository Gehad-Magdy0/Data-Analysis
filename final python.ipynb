```python
df = order_items.merge(orders, on='order_id') \
 .merge(products, on='product_id') \
 .merge(customers, on='customer_id') \
 .merge(brands, on='brand_id') \
 .merge(categories, on='category_id')
```

```python
staffs_with_stores = staffs.merge(
    stores[['store_id', 'store_name', 'city']], 
    on='store_id',
    how='left' 
)
print(" Merged Staffs and Stores Table ")
display(staffs_with_stores.head())
print("\nColumns in the Merged Table")
print(staffs_with_stores.columns.tolist())
```

```python
order_items.groupby('discount').size().plot(kind='pie' ,autopct='%1.1f%%'  , title = 'Discount rating')
```

```python
df.groupby('category_name').size().plot(kind='barh',  title='Number of Products per Category',grid=True)
```

```python
df['total_price'] = df['quantity'] * df['list_price_x']
top_products = df.groupby('product_name')['total_price'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(14,6))
top_products.plot(kind='bar', color='purple', grid=True)
plt.title('Top 10 Best-Selling Products', fontsize=16)
plt.xlabel('Product Name')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby('category_name')['total_price'].sum()
plt.figure(figsize=(10,8))
category_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales ber Category')
plt.ylabel('')
plt.show()
```

```python
qty_cat = df.groupby("category_name")["quantity"].sum()
plt.figure(figsize=(9,4))
sns.barplot(x=qty_cat.index, y=qty_cat.values)
plt.xticks(rotation=45)
plt.title("Total Quantity Sold by Category")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(df["list_price_x"], kde=True)
plt.title("Price Distribution")
plt.show()
```

```python
plt.figure(figsize=(9,4))
df["brand_name"].value_counts().plot(kind="barh")
plt.title("Sales per Brand")
plt.xticks(rotation=45)
plt.show()
```

```python
df["order_month"] = pd.to_datetime(df["order_date"]).dt.to_period("M")
monthly = df.groupby("order_month")["order_id"].nunique()
monthly.plot(figsize=(10,4))
plt.title("Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.show()
```

```python
cat_counts = df["category_name"].value_counts()
plt.figure(figsize=(7,4))
sns.barplot(x=cat_counts.index, y=cat_counts.values)
plt.xticks(rotation=45)
plt.title("Orders by Category")
plt.show()
```

```python
df["revenue"] = df["list_price_x"] * df["quantity"]
cat_rev = df.groupby("category_name")["revenue"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,4))
sns.barplot(x=cat_rev.index, y=cat_rev.values)
plt.xticks(rotation=45)
plt.title("Revenue per Category")
plt.show()
```

```python
brand_rev = df.groupby("brand_name")["revenue"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,4))
sns.barplot(x=brand_rev.index, y=brand_rev.values)
plt.xticks(rotation=45)
plt.title("Revenue per Brand")
plt.show()
```

```python
state_counts = df["state"].value_counts()
plt.figure(figsize=(8,4))
sns.barplot(x=state_counts.index, y=state_counts.values)
plt.xticks(rotation=90)
plt.title("Customers by State")
plt.show()
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
df["required_date"] = pd.to_datetime(df["required_date"])
df["delay"] = (df["required_date"] - df["order_date"]).dt.days
sns.histplot(df["delay"], kde=True)
plt.title("Shipping Delay")
plt.show()
```

```python
top_customers = df.groupby("customer_id")["revenue"].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(8,4))
sns.barplot(x=top_customers.index, y=top_customers.values)
plt.title("Top 10 Customers by Revenue")
plt.show()
```

```python
city_counts = df["city"].value_counts().head(15)
plt.figure(figsize=(10,4))
sns.barplot(x=city_counts.index, y=city_counts.values)
plt.xticks(rotation=90)
plt.title("Top Cities by Number of Customers")
plt.show()
```
